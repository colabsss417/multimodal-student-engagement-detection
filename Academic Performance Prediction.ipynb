```python
# ============================================================
# STEP 3: DATA PREPROCESSING
# MIN-MAX NORMALIZATION + ONE-HOT ENCODING
# ============================================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import MinMaxScaler, OneHotEncoder


# ============================================================
# 1. CREATE A COPY OF MERGED DATASET
# ============================================================

preprocessed_df = merged_df.copy()

print("=" * 70)
print("STEP 3: DATA PREPROCESSING")
print("=" * 70)

print("\nOriginal Dataset Shape:", preprocessed_df.shape)


# ============================================================
# 2. DEFINE NUMERICAL FEATURES
# ============================================================

numerical_columns = [

    "age",

    "structured_attention_score",

    "session_duration",

    "text_attention_score",

    "engagement_score",

    "text_length"

]


print("\nNumerical Features:")

for column in numerical_columns:

    print("-", column)


# ============================================================
# 3. DISPLAY VALUES BEFORE NORMALIZATION
# ============================================================

print("\n" + "=" * 70)
print("NUMERICAL FEATURES BEFORE NORMALIZATION")
print("=" * 70)

print(

    preprocessed_df[
        numerical_columns
    ].head()

)


# ============================================================
# 4. INITIALIZE MIN-MAX SCALER
# ============================================================

scaler = MinMaxScaler(
    feature_range=(0, 1)
)


# ============================================================
# 5. APPLY MIN-MAX NORMALIZATION
# ============================================================

preprocessed_df[
    numerical_columns
] = scaler.fit_transform(

    preprocessed_df[
        numerical_columns
    ]

)


# ============================================================
# 6. DISPLAY NORMALIZED VALUES
# ============================================================

print("\n" + "=" * 70)
print("NUMERICAL FEATURES AFTER MIN-MAX NORMALIZATION")
print("=" * 70)

print(

    preprocessed_df[
        numerical_columns
    ].head()

)


# ============================================================
# 7. VERIFY NORMALIZATION RANGE
# ============================================================

print("\n" + "=" * 70)
print("NORMALIZATION RANGE VERIFICATION")
print("=" * 70)


for column in numerical_columns:

    minimum = preprocessed_df[column].min()

    maximum = preprocessed_df[column].max()

    print(
        f"{column:<30} "
        f"Min = {minimum:.4f} | "
        f"Max = {maximum:.4f}"
    )


# ============================================================
# 8. DEFINE CATEGORICAL INPUT FEATURES
# ============================================================

categorical_columns = [

    "behavior",

    "gender",

    "device"

]


print("\n" + "=" * 70)
print("CATEGORICAL FEATURES")
print("=" * 70)


for column in categorical_columns:

    print(
        f"\n{column}:"
    )

    print(
        preprocessed_df[
            column
        ].unique()
    )


# ============================================================
# 9. APPLY ONE-HOT ENCODING
# ============================================================

preprocessed_df = pd.get_dummies(

    preprocessed_df,

    columns=categorical_columns,

    dtype=int

)


# ============================================================
# 10. DISPLAY ENCODED FEATURES
# ============================================================

print("\n" + "=" * 70)
print("ONE-HOT ENCODED FEATURES")
print("=" * 70)


encoded_columns = [

    column

    for column in preprocessed_df.columns

    if any(

        column.startswith(
            category + "_"
        )

        for category in categorical_columns

    )

]


for column in encoded_columns:

    print("-", column)


# ============================================================
# 11. ENCODE TARGET COLUMN
# ============================================================

target_mapping = {

    "Not Engaged": 0,

    "Engaged": 1

}


preprocessed_df[
    "engagement"
] = preprocessed_df[
    "engagement"
].map(target_mapping)


# ============================================================
# 12. VERIFY TARGET ENCODING
# ============================================================

print("\n" + "=" * 70)
print("TARGET ENCODING")
print("=" * 70)


print(

    preprocessed_df[
        "engagement"
    ].value_counts()

)


print("\nTarget Mapping:")

print("Not Engaged -> 0")

print("Engaged     -> 1")


# ============================================================
# 13. DISPLAY FINAL COLUMNS
# ============================================================

print("\n" + "=" * 70)
print("PREPROCESSED DATASET COLUMNS")
print("=" * 70)


for number, column in enumerate(

    preprocessed_df.columns,

    start=1

):

    print(
        f"{number}. {column}"
    )


# ============================================================
# 14. DISPLAY FIRST FIVE RECORDS
# ============================================================

print("\n" + "=" * 70)
print("FIRST 5 PREPROCESSED RECORDS")
print("=" * 70)


print(

    preprocessed_df.head()

)


# ============================================================
# 15. CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUE VERIFICATION")
print("=" * 70)


print(

    "Total Missing Values:",

    preprocessed_df
    .isnull()
    .sum()
    .sum()

)


# ============================================================
# 16. CHECK FINAL DATASET SHAPE
# ============================================================

print("\n" + "=" * 70)
print("FINAL PREPROCESSING SUMMARY")
print("=" * 70)


print(

    "Dataset Shape Before Preprocessing:",

    merged_df.shape

)


print(

    "Dataset Shape After Preprocessing :",

    preprocessed_df.shape

)


print(

    "Numerical Features Normalized     :",

    len(numerical_columns)

)


print(

    "Categorical Features Encoded      :",

    len(categorical_columns)

)


print(

    "One-Hot Features Created          :",

    len(encoded_columns)

)


print(

    "Target Column                     : engagement"

)


print("\nSTEP 3 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 4: IMAGE PREPROCESSING
# IMAGE MATCHING + RESIZING + NORMALIZATION + AUGMENTATION
# ============================================================

import os
import numpy as np
import pandas as pd
import tensorflow as tf

from tensorflow.keras.preprocessing.image import (
    load_img,
    img_to_array
)

from tensorflow.keras.preprocessing.image import ImageDataGenerator


# ============================================================
# 1. DEFINE IMAGE FOLDER PATH
# ============================================================

IMAGE_FOLDER_PATH = r"E:\Projects\2_Enterprises\Images"


# ============================================================
# 2. FIND ALL IMAGE FILES RECURSIVELY
# ============================================================

image_path_dictionary = {}

valid_extensions = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp"
)

for root, directories, files in os.walk(IMAGE_FOLDER_PATH):

    for file in files:

        if file.lower().endswith(valid_extensions):

            image_path_dictionary[file] = os.path.join(
                root,
                file
            )


print("=" * 70)
print("STEP 4: IMAGE PREPROCESSING")
print("=" * 70)

print("\nTotal Actual Images Found:",
      len(image_path_dictionary))


print("\nFirst 10 Image Files:")

for i, (name, path) in enumerate(
    image_path_dictionary.items()
):

    if i == 10:
        break

    print(name, "->", path)


# ============================================================
# 3. MATCH DATASET IMAGE NAMES WITH ACTUAL IMAGES
# ============================================================

preprocessed_df["image_path"] = (
    preprocessed_df["image_name"]
    .map(image_path_dictionary)
)


print("\n" + "=" * 70)
print("IMAGE MATCHING RESULT")
print("=" * 70)

print(
    "Total Dataset Records :",
    len(preprocessed_df)
)

print(
    "Matched Images        :",
    preprocessed_df["image_path"].notna().sum()
)

print(
    "Missing Images        :",
    preprocessed_df["image_path"].isna().sum()
)


# ============================================================
# 4. DISPLAY MISSING IMAGE NAMES
# ============================================================

missing_images = (

    preprocessed_df.loc[
        preprocessed_df["image_path"].isna(),
        "image_name"
    ]

    .unique()

)


print("\nNumber of Unique Missing Images:",
      len(missing_images))


if len(missing_images) > 0:

    print("\nFirst 20 Missing Image Names:")

    print(missing_images[:20])


# ============================================================
# 5. KEEP ONLY RECORDS WITH AVAILABLE IMAGES
# ============================================================

image_df = (

    preprocessed_df
    .dropna(subset=["image_path"])
    .reset_index(drop=True)

)


print("\nDataset Shape After Image Matching:",
      image_df.shape)


# ============================================================
# 6. DEFINE IMAGE DIMENSIONS
# ============================================================

IMG_HEIGHT = 224

IMG_WIDTH = 224

IMG_CHANNELS = 3


# ============================================================
# 7. FUNCTION FOR IMAGE LOADING AND RESIZING
# ============================================================

def preprocess_image(image_path):

    try:

        # Load RGB image and resize to 224 x 224
        image = load_img(

            image_path,

            target_size=(
                IMG_HEIGHT,
                IMG_WIDTH
            ),

            color_mode="rgb"

        )


        # Convert image into numerical array
        image = img_to_array(image)


        # Convert datatype
        image = image.astype("float32")


        # Pixel normalization [0,255] -> [0,1]
        image = image / 255.0


        return image


    except Exception as error:

        print(
            "Error processing:",
            image_path,
            error
        )

        return None


# ============================================================
# 8. LOAD AND PREPROCESS ALL IMAGES
# ============================================================

processed_images = []

valid_indices = []


print("\n" + "=" * 70)
print("LOADING AND PREPROCESSING IMAGES")
print("=" * 70)


for index, image_path in enumerate(
    image_df["image_path"]
):

    processed_image = preprocess_image(
        image_path
    )


    if processed_image is not None:

        processed_images.append(
            processed_image
        )

        valid_indices.append(
            index
        )


# ============================================================
# 9. CONVERT IMAGE LIST INTO NUMPY ARRAY
# ============================================================

X_images = np.array(

    processed_images,

    dtype=np.float32

)


# Keep only successfully processed records
image_df = (

    image_df
    .iloc[valid_indices]
    .reset_index(drop=True)

)


print("\n" + "=" * 70)
print("IMAGE ARRAY DETAILS")
print("=" * 70)


print(
    "Image Array Shape:",
    X_images.shape
)


print(
    "Image Data Type:",
    X_images.dtype
)


print(
    "Minimum Pixel Value:",
    X_images.min()
)


print(
    "Maximum Pixel Value:",
    X_images.max()
)


# ============================================================
# 10. DEFINE IMAGE AUGMENTATION
# ============================================================

image_augmentation = ImageDataGenerator(

    rotation_range=15,

    horizontal_flip=True,

    zoom_range=0.1,

    width_shift_range=0.1,

    height_shift_range=0.1,

    fill_mode="nearest"

)


print("\n" + "=" * 70)
print("IMAGE AUGMENTATION SETTINGS")
print("=" * 70)


print("Random Rotation     : ±15 degrees")

print("Horizontal Flipping : Enabled")

print("Random Zoom         : Approximately 0.9–1.1")

print("Width Shift         : 10%")

print("Height Shift        : 10%")


# ============================================================
# 11. CREATE AUGMENTATION GENERATOR
# ============================================================

augmentation_generator = image_augmentation.flow(

    X_images,

    batch_size=32,

    shuffle=True

)


# ============================================================
# 12. GENERATE ONE AUGMENTED BATCH FOR VERIFICATION
# ============================================================

augmented_batch = next(
    augmentation_generator
)


print("\n" + "=" * 70)
print("AUGMENTATION VERIFICATION")
print("=" * 70)


print(
    "Original Image Array Shape:",
    X_images.shape
)


print(
    "Augmented Batch Shape:",
    augmented_batch.shape
)


print(
    "Augmented Minimum Value:",
    augmented_batch.min()
)


print(
    "Augmented Maximum Value:",
    augmented_batch.max()
)


# ============================================================
# 13. EXTRACT TARGET LABELS
# ============================================================

y = image_df[
    "engagement"
].values.astype("int32")


print("\n" + "=" * 70)
print("TARGET ARRAY")
print("=" * 70)


print(
    "Target Shape:",
    y.shape
)


print(
    "Target Classes:",
    np.unique(y)
)


print(
    "Class Distribution:"
)


unique_classes, class_counts = np.unique(

    y,

    return_counts=True

)


for class_value, count in zip(

    unique_classes,

    class_counts

):

    print(
        f"Class {class_value}: {count}"
    )


# ============================================================
# 14. FINAL IMAGE PREPROCESSING SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL IMAGE PREPROCESSING SUMMARY")
print("=" * 70)


print(
    "Final Dataset Records :",
    len(image_df)
)


print(
    "Processed Image Shape :",
    X_images.shape
)


print(
    "Target Shape          :",
    y.shape
)


print(
    "Image Height          :",
    IMG_HEIGHT
)


print(
    "Image Width           :",
    IMG_WIDTH
)


print(
    "Image Channels        :",
    IMG_CHANNELS
)


print(
    "Pixel Value Range     : [0, 1]"
)


print(
    "Augmentation          : Configured"
)


print("\nSTEP 4 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 10: TRANSFORMER ENCODER + BiLSTM MODEL
# ============================================================

import numpy as np
import tensorflow as tf

from tensorflow.keras.models import Model

from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    MultiHeadAttention,
    LayerNormalization,
    Add,
    Bidirectional,
    LSTM
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

from tensorflow.keras.optimizers import Adam


# ============================================================
# 1. SET RANDOM SEEDS
# ============================================================

np.random.seed(42)

tf.random.set_seed(42)


# ============================================================
# 2. DISPLAY INPUT DATA DETAILS
# ============================================================

print("=" * 70)
print("STEP 10: TRANSFORMER ENCODER + BiLSTM MODEL")
print("=" * 70)

print("\nTraining Sequence Shape   :", X_train_sequence.shape)

print("Validation Sequence Shape :", X_validation_sequence.shape)

print("Testing Sequence Shape    :", X_test_sequence.shape)

print("\nTraining Target Shape     :", y_train.shape)

print("Validation Target Shape   :", y_validation.shape)

print("Testing Target Shape      :", y_test.shape)


# ============================================================
# 3. DEFINE MODEL PARAMETERS
# ============================================================

SEQUENCE_LENGTH = X_train_sequence.shape[1]

INPUT_DIMENSION = X_train_sequence.shape[2]


# Transformer parameters

NUM_HEADS = 4

KEY_DIMENSION = 16

FEED_FORWARD_DIMENSION = 64

TRANSFORMER_DROPOUT = 0.2


# BiLSTM parameters

BILSTM_UNITS = 64

BILSTM_DROPOUT = 0.3


# Dense layer parameters

DENSE_UNITS = 64

DENSE_DROPOUT = 0.3


# Training parameters

LEARNING_RATE = 0.001

BATCH_SIZE = 32

EPOCHS = 50


print("\n" + "=" * 70)
print("MODEL CONFIGURATION")
print("=" * 70)

print("Sequence Length          :", SEQUENCE_LENGTH)

print("Input Dimension          :", INPUT_DIMENSION)

print("Attention Heads          :", NUM_HEADS)

print("Attention Key Dimension  :", KEY_DIMENSION)

print("Feed Forward Dimension   :", FEED_FORWARD_DIMENSION)

print("BiLSTM Units             :", BILSTM_UNITS)

print("Dense Units              :", DENSE_UNITS)

print("Learning Rate            :", LEARNING_RATE)

print("Batch Size               :", BATCH_SIZE)

print("Maximum Epochs           :", EPOCHS)


# ============================================================
# 4. TRANSFORMER ENCODER FUNCTION
# ============================================================

def transformer_encoder(

    inputs,

    num_heads,

    key_dimension,

    feed_forward_dimension,

    dropout_rate

):


    # --------------------------------------------------------
    # MULTI-HEAD SELF-ATTENTION
    # --------------------------------------------------------

    attention_output = MultiHeadAttention(

        num_heads=num_heads,

        key_dim=key_dimension,

        dropout=dropout_rate,

        name="multi_head_self_attention"

    )(

        inputs,

        inputs

    )


    # --------------------------------------------------------
    # ATTENTION DROPOUT
    # --------------------------------------------------------

    attention_output = Dropout(

        dropout_rate,

        name="attention_dropout"

    )(attention_output)


    # --------------------------------------------------------
    # FIRST RESIDUAL CONNECTION
    # --------------------------------------------------------

    residual_attention = Add(

        name="attention_residual_connection"

    )([

        inputs,

        attention_output

    ])


    # --------------------------------------------------------
    # FIRST LAYER NORMALIZATION
    # --------------------------------------------------------

    attention_normalized = LayerNormalization(

        epsilon=1e-6,

        name="attention_layer_normalization"

    )(residual_attention)


    # --------------------------------------------------------
    # FEED FORWARD NETWORK
    # --------------------------------------------------------

    feed_forward_output = Dense(

        feed_forward_dimension,

        activation="relu",

        name="transformer_feed_forward_1"

    )(attention_normalized)


    feed_forward_output = Dropout(

        dropout_rate,

        name="feed_forward_dropout"

    )(feed_forward_output)


    # Project back to original input dimension
    feed_forward_output = Dense(

        inputs.shape[-1],

        name="transformer_feed_forward_2"

    )(feed_forward_output)


    # --------------------------------------------------------
    # SECOND RESIDUAL CONNECTION
    # --------------------------------------------------------

    residual_feed_forward = Add(

        name="feed_forward_residual_connection"

    )([

        attention_normalized,

        feed_forward_output

    ])


    # --------------------------------------------------------
    # SECOND LAYER NORMALIZATION
    # --------------------------------------------------------

    transformer_output = LayerNormalization(

        epsilon=1e-6,

        name="feed_forward_layer_normalization"

    )(residual_feed_forward)


    return transformer_output


# ============================================================
# 5. DEFINE INPUT LAYER
# ============================================================

model_input = Input(

    shape=(

        SEQUENCE_LENGTH,

        INPUT_DIMENSION

    ),

    name="multimodal_sequence_input"

)


# ============================================================
# 6. TRANSFORMER ENCODER
# ============================================================

transformer_output = transformer_encoder(

    model_input,

    num_heads=NUM_HEADS,

    key_dimension=KEY_DIMENSION,

    feed_forward_dimension=FEED_FORWARD_DIMENSION,

    dropout_rate=TRANSFORMER_DROPOUT

)


# ============================================================
# 7. BIDIRECTIONAL LSTM
# ============================================================

bilstm_output = Bidirectional(

    LSTM(

        units=BILSTM_UNITS,

        return_sequences=False,

        dropout=BILSTM_DROPOUT

    ),

    name="bidirectional_lstm"

)(transformer_output)


# ============================================================
# 8. DENSE CLASSIFICATION LAYER
# ============================================================

dense_output = Dense(

    units=DENSE_UNITS,

    activation="relu",

    name="classification_dense"

)(bilstm_output)


# ============================================================
# 9. DROPOUT
# ============================================================

dense_output = Dropout(

    DENSE_DROPOUT,

    name="classification_dropout"

)(dense_output)


# ============================================================
# 10. BINARY OUTPUT LAYER
# ============================================================

model_output = Dense(

    units=1,

    activation="sigmoid",

    name="engagement_prediction"

)(dense_output)


# ============================================================
# 11. CREATE FINAL MODEL
# ============================================================

model = Model(

    inputs=model_input,

    outputs=model_output,

    name="Transformer_BiLSTM_Model"

)


# ============================================================
# 12. DISPLAY MODEL ARCHITECTURE
# ============================================================

print("\n" + "=" * 70)
print("TRANSFORMER + BiLSTM MODEL ARCHITECTURE")
print("=" * 70)


model.summary()


# ============================================================
# 13. COMPILE MODEL
# ============================================================

optimizer = Adam(

    learning_rate=LEARNING_RATE

)


model.compile(

    optimizer=optimizer,

    loss="binary_crossentropy",

    metrics=[

        "accuracy",

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )

    ]

)


print("\n" + "=" * 70)
print("MODEL COMPILATION")
print("=" * 70)

print("Optimizer : Adam")

print("Loss      : Binary Crossentropy")

print("Metrics   : Accuracy, Precision, Recall, AUC")


# ============================================================
# 14. EARLY STOPPING
# ============================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=10,

    mode="min",

    restore_best_weights=True,

    verbose=1

)


# ============================================================
# 15. REDUCE LEARNING RATE
# ============================================================

reduce_learning_rate = ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=5,

    min_lr=1e-6,

    mode="min",

    verbose=1

)


# ============================================================
# 16. SAVE BEST MODEL
# ============================================================

model_checkpoint = ModelCheckpoint(

    filepath="best_transformer_bilstm_model.keras",

    monitor="val_loss",

    save_best_only=True,

    mode="min",

    verbose=1

)


# ============================================================
# 17. CALLBACK LIST
# ============================================================

callbacks = [

    early_stopping,

    reduce_learning_rate,

    model_checkpoint

]


# ============================================================
# 18. TRAIN MODEL
# ============================================================

print("\n" + "=" * 70)
print("STARTING MODEL TRAINING")
print("=" * 70)


history = model.fit(

    X_train_sequence,

    y_train,

    validation_data=(

        X_validation_sequence,

        y_validation

    ),

    epochs=EPOCHS,

    batch_size=BATCH_SIZE,

    callbacks=callbacks,

    verbose=1

)


# ============================================================
# 19. DISPLAY TRAINING COMPLETION
# ============================================================

print("\n" + "=" * 70)
print("MODEL TRAINING COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 11: 5-FOLD STRATIFIED GROUP CROSS-VALIDATION
# TRANSFORMER + BiLSTM MODEL
# ============================================================

import numpy as np
import pandas as pd
import tensorflow as tf
import gc

from sklearn.model_selection import StratifiedGroupKFold

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

from tensorflow.keras.models import Model

from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    MultiHeadAttention,
    LayerNormalization,
    Add,
    Bidirectional,
    LSTM
)

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)


# ============================================================
# 1. SET RANDOM SEEDS
# ============================================================

np.random.seed(42)

tf.random.set_seed(42)


# ============================================================
# 2. PREPARE COMPLETE DATASET
# ============================================================

X_cv = fused_features.astype("float32")

y_cv = y.astype("int32")

groups_cv = image_df[
    "image_name"
].values


print("=" * 70)
print("5-FOLD STRATIFIED GROUP CROSS-VALIDATION")
print("=" * 70)

print("\nFeature Shape :", X_cv.shape)

print("Target Shape  :", y_cv.shape)

print("Groups Shape  :", groups_cv.shape)

print(
    "Unique Groups :",
    len(np.unique(groups_cv))
)


# ============================================================
# 3. RESHAPE COMPLETE FEATURE DATA
# ============================================================

SEQUENCE_LENGTH = 16

FEATURES_PER_STEP = 16


X_cv_sequence = X_cv.reshape(

    X_cv.shape[0],

    SEQUENCE_LENGTH,

    FEATURES_PER_STEP

)


print(
    "\nSequence Feature Shape:",
    X_cv_sequence.shape
)


# ============================================================
# 4. VERIFY ALIGNMENT
# ============================================================

assert (

    X_cv_sequence.shape[0]

    ==

    y_cv.shape[0]

    ==

    groups_cv.shape[0]

)


print("\nFeature-Target-Group Alignment: CORRECT")


# ============================================================
# 5. MODEL BUILDING FUNCTION
# ============================================================

def build_transformer_bilstm_model():


    # --------------------------------------------------------
    # INPUT LAYER
    # --------------------------------------------------------

    inputs = Input(

        shape=(

            SEQUENCE_LENGTH,

            FEATURES_PER_STEP

        )

    )


    # --------------------------------------------------------
    # MULTI-HEAD SELF-ATTENTION
    # --------------------------------------------------------

    attention_output = MultiHeadAttention(

        num_heads=4,

        key_dim=16,

        dropout=0.2

    )(

        inputs,

        inputs

    )


    # --------------------------------------------------------
    # ATTENTION DROPOUT
    # --------------------------------------------------------

    attention_output = Dropout(

        0.2

    )(attention_output)


    # --------------------------------------------------------
    # FIRST RESIDUAL CONNECTION
    # --------------------------------------------------------

    x = Add()(

        [

            inputs,

            attention_output

        ]

    )


    # --------------------------------------------------------
    # FIRST LAYER NORMALIZATION
    # --------------------------------------------------------

    x = LayerNormalization(

        epsilon=1e-6

    )(x)


    # --------------------------------------------------------
    # TRANSFORMER FEED FORWARD NETWORK
    # --------------------------------------------------------

    feed_forward = Dense(

        64,

        activation="relu"

    )(x)


    feed_forward = Dropout(

        0.2

    )(feed_forward)


    feed_forward = Dense(

        FEATURES_PER_STEP

    )(feed_forward)


    # --------------------------------------------------------
    # SECOND RESIDUAL CONNECTION
    # --------------------------------------------------------

    x = Add()(

        [

            x,

            feed_forward

        ]

    )


    # --------------------------------------------------------
    # SECOND LAYER NORMALIZATION
    # --------------------------------------------------------

    x = LayerNormalization(

        epsilon=1e-6

    )(x)


    # --------------------------------------------------------
    # BIDIRECTIONAL LSTM
    # --------------------------------------------------------

    x = Bidirectional(

        LSTM(

            units=64,

            return_sequences=False,

            dropout=0.3

        )

    )(x)


    # --------------------------------------------------------
    # DENSE LAYER
    # --------------------------------------------------------

    x = Dense(

        64,

        activation="relu"

    )(x)


    # --------------------------------------------------------
    # DROPOUT
    # --------------------------------------------------------

    x = Dropout(

        0.3

    )(x)


    # --------------------------------------------------------
    # OUTPUT LAYER
    # --------------------------------------------------------

    outputs = Dense(

        1,

        activation="sigmoid"

    )(x)


    # --------------------------------------------------------
    # CREATE MODEL
    # --------------------------------------------------------

    cv_model = Model(

        inputs=inputs,

        outputs=outputs

    )


    # --------------------------------------------------------
    # COMPILE MODEL
    # --------------------------------------------------------

    cv_model.compile(

        optimizer=Adam(

            learning_rate=0.001

        ),

        loss="binary_crossentropy",

        metrics=["accuracy"]

    )


    return cv_model


# ============================================================
# 6. INITIALIZE STRATIFIED GROUP 5-FOLD CV
# ============================================================

sgkf = StratifiedGroupKFold(

    n_splits=5,

    shuffle=True,

    random_state=42

)


# ============================================================
# 7. CREATE RESULT STORAGE
# ============================================================

fold_results = []


# ============================================================
# 8. START 5-FOLD CROSS-VALIDATION
# ============================================================

print("\n" + "=" * 70)

print("STARTING 5-FOLD CROSS-VALIDATION")

print("=" * 70)


for fold, (

    train_index,

    validation_index

) in enumerate(

    sgkf.split(

        X_cv_sequence,

        y_cv,

        groups=groups_cv

    ),

    start=1

):


    print("\n\n" + "=" * 70)

    print(f"FOLD {fold} / 5")

    print("=" * 70)


    # ========================================================
    # CLEAR PREVIOUS MODEL SESSION
    # ========================================================

    tf.keras.backend.clear_session()

    gc.collect()


    # ========================================================
    # CREATE FOLD DATA
    # ========================================================

    X_fold_train = X_cv_sequence[
        train_index
    ]

    X_fold_validation = X_cv_sequence[
        validation_index
    ]


    y_fold_train = y_cv[
        train_index
    ]

    y_fold_validation = y_cv[
        validation_index
    ]


    train_groups = groups_cv[
        train_index
    ]

    validation_groups = groups_cv[
        validation_index
    ]


    # ========================================================
    # DISPLAY FOLD DETAILS
    # ========================================================

    print(
        "\nTraining Shape   :",
        X_fold_train.shape
    )

    print(
        "Validation Shape :",
        X_fold_validation.shape
    )


    print(
        "\nTraining Groups   :",
        len(np.unique(train_groups))
    )

    print(
        "Validation Groups :",
        len(np.unique(validation_groups))
    )


    # ========================================================
    # CHECK GROUP LEAKAGE
    # ========================================================

    group_overlap = (

        set(train_groups)

        &

        set(validation_groups)

    )


    print(
        "Group Overlap     :",
        len(group_overlap)
    )


    assert len(group_overlap) == 0


    # ========================================================
    # DISPLAY CLASS DISTRIBUTION
    # ========================================================

    print(
        "\nTraining Class Distribution:"
    )

    print(

        np.bincount(
            y_fold_train
        )

    )


    print(
        "Validation Class Distribution:"
    )

    print(

        np.bincount(
            y_fold_validation
        )

    )


    # ========================================================
    # BUILD NEW MODEL
    # ========================================================

    cv_model = build_transformer_bilstm_model()


    print(
        "\nNew Transformer-BiLSTM Model Created"
    )


    # ========================================================
    # DEFINE CALLBACKS
    # ========================================================

    early_stopping = EarlyStopping(

        monitor="val_loss",

        patience=8,

        restore_best_weights=True,

        mode="min",

        verbose=0

    )


    reduce_lr = ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.5,

        patience=4,

        min_lr=1e-6,

        mode="min",

        verbose=0

    )


    # ========================================================
    # TRAIN MODEL
    # ========================================================

    print(
        f"\nTraining Fold {fold}..."
    )


    fold_history = cv_model.fit(

        X_fold_train,

        y_fold_train,

        validation_data=(

            X_fold_validation,

            y_fold_validation

        ),

        epochs=50,

        batch_size=32,

        callbacks=[

            early_stopping,

            reduce_lr

        ],

        verbose=0

    )


    print(
        "Epochs Trained:",
        len(fold_history.history["loss"])
    )


    # ========================================================
    # GENERATE PROBABILITY PREDICTIONS
    # ========================================================

    y_fold_probability = cv_model.predict(

        X_fold_validation,

        batch_size=32,

        verbose=0

    ).ravel()


    # ========================================================
    # CONVERT PROBABILITY INTO CLASS
    # ========================================================

    y_fold_prediction = (

        y_fold_probability >= 0.5

    ).astype("int32")


    # ========================================================
    # CALCULATE CLASSIFICATION METRICS
    # ========================================================

    fold_accuracy = accuracy_score(

        y_fold_validation,

        y_fold_prediction

    )


    fold_precision = precision_score(

        y_fold_validation,

        y_fold_prediction,

        zero_division=0

    )


    fold_recall = recall_score(

        y_fold_validation,

        y_fold_prediction,

        zero_division=0

    )


    fold_f1 = f1_score(

        y_fold_validation,

        y_fold_prediction,

        zero_division=0

    )


    fold_auc = roc_auc_score(

        y_fold_validation,

        y_fold_probability

    )


    # ========================================================
    # CALCULATE ERROR METRICS
    # ========================================================

    fold_mse = mean_squared_error(

        y_fold_validation,

        y_fold_probability

    )


    fold_mae = mean_absolute_error(

        y_fold_validation,

        y_fold_probability

    )


    fold_rmse = np.sqrt(

        fold_mse

    )


    fold_r2 = r2_score(

        y_fold_validation,

        y_fold_probability

    )


    # ========================================================
    # STORE RESULTS
    # ========================================================

    fold_results.append({

        "Fold": fold,

        "Accuracy": fold_accuracy,

        "Precision": fold_precision,

        "Recall": fold_recall,

        "F1-Score": fold_f1,

        "AUC": fold_auc,

        "MSE": fold_mse,

        "MAE": fold_mae,

        "RMSE": fold_rmse,

        "R2": fold_r2

    })


    # ========================================================
    # DISPLAY FOLD RESULTS
    # ========================================================

    print("\n" + "-" * 70)

    print(f"FOLD {fold} RESULTS")

    print("-" * 70)


    print(
        f"Accuracy  : {fold_accuracy:.4f}"
    )

    print(
        f"Precision : {fold_precision:.4f}"
    )

    print(
        f"Recall    : {fold_recall:.4f}"
    )

    print(
        f"F1-Score  : {fold_f1:.4f}"
    )

    print(
        f"AUC       : {fold_auc:.4f}"
    )

    print(
        f"MSE       : {fold_mse:.4f}"
    )

    print(
        f"MAE       : {fold_mae:.4f}"
    )

    print(
        f"RMSE      : {fold_rmse:.4f}"
    )

    print(
        f"R2        : {fold_r2:.4f}"
    )


# ============================================================
# 9. CONVERT RESULTS TO DATAFRAME
# ============================================================

cv_results_df = pd.DataFrame(

    fold_results

)


# ============================================================
# 10. DISPLAY ALL FOLD RESULTS
# ============================================================

print("\n\n" + "=" * 90)

print("5-FOLD CROSS-VALIDATION RESULTS")

print("=" * 90)


print(

    cv_results_df.to_string(

        index=False

    )

)


# ============================================================
# 11. CALCULATE MEAN RESULTS
# ============================================================

mean_results = (

    cv_results_df

    .drop(columns=["Fold"])

    .mean()

)


# ============================================================
# 12. CALCULATE STANDARD DEVIATION
# ============================================================

std_results = (

    cv_results_df

    .drop(columns=["Fold"])

    .std()

)

# ============================================================
# 14. SAVE RESULTS
# ============================================================

cv_results_df.to_csv(

    "5_fold_cross_validation_results.csv",

    index=False

)


print("\nResults Saved As:")

print("5_fold_cross_validation_results.csv")


print("\n5-FOLD CROSS-VALIDATION COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 10: ABLATION STUDY
# TO–TBi-STMNet MULTIMODAL MODEL
# ============================================================

import numpy as np
import pandas as pd
import time
import warnings

warnings.filterwarnings("ignore")

import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import accuracy_score

from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    LSTM,
    Bidirectional,
    MultiHeadAttention,
    LayerNormalization,
    Add,
    GlobalAveragePooling1D
)
from tensorflow.keras.callbacks import EarlyStopping


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)


# ============================================================
# 2. PREPARE TARGET
# ============================================================

y_ablation = np.asarray(y).ravel()

encoder = LabelEncoder()
y_ablation = encoder.fit_transform(y_ablation)

num_classes = len(np.unique(y_ablation))


# ============================================================
# 3. PREPARE MODALITIES
# ============================================================

X_image = np.asarray(cnn_features, dtype=np.float32)
X_text = np.asarray(text_features, dtype=np.float32)
X_multimodal = np.asarray(fused_features, dtype=np.float32)

print("=" * 90)
print("ABLATION STUDY - TO–TBi-STMNet")
print("=" * 90)

print("Image Features       :", X_image.shape)
print("Text Features       :", X_text.shape)
print("Multimodal Features :", X_multimodal.shape)
print("Target Shape        :", y_ablation.shape)


# ============================================================
# 4. TRAIN-TEST SPLIT
# ============================================================

indices = np.arange(len(y_ablation))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.20,
    random_state=SEED,
    stratify=y_ablation
)


# ============================================================
# 5. ABLATION RESULT STORAGE
# ============================================================

ablation_results = []


# ============================================================
# 6. PREPARE DATA FUNCTION
# ============================================================

def prepare_data(X):

    scaler = StandardScaler()

    X_train = scaler.fit_transform(X[train_idx]).astype(np.float32)
    X_test = scaler.transform(X[test_idx]).astype(np.float32)

    X_train_seq = X_train.reshape(
        X_train.shape[0],
        X_train.shape[1],
        1
    )

    X_test_seq = X_test.reshape(
        X_test.shape[0],
        X_test.shape[1],
        1
    )

    return X_train_seq, X_test_seq


# ============================================================
# 7. TRANSFORMER-ONLY MODEL
# ============================================================

def build_transformer_only(input_shape):

    inputs = Input(shape=input_shape)

    attention = MultiHeadAttention(
        num_heads=4,
        key_dim=16
    )(inputs, inputs)

    x = Add()([inputs, attention])

    x = LayerNormalization()(x)

    x = GlobalAveragePooling1D()(x)

    x = Dense(128, activation="relu")(x)

    x = Dropout(0.3)(x)

    outputs = Dense(
        num_classes,
        activation="softmax"
    )(x)

    model = Model(inputs, outputs)

    return model


# ============================================================
# 8. BiLSTM-ONLY MODEL
# ============================================================

def build_bilstm_only(input_shape):

    model = Sequential([
        Input(shape=input_shape),

        Bidirectional(
            LSTM(64)
        ),

        Dense(128, activation="relu"),

        Dropout(0.3),

        Dense(
            num_classes,
            activation="softmax"
        )
    ])

    return model


# ============================================================
# 9. TRANSFORMER + BiLSTM MODEL
# ============================================================

def build_transformer_bilstm(input_shape):

    inputs = Input(shape=input_shape)

    x = Bidirectional(
        LSTM(64, return_sequences=True)
    )(inputs)

    attention = MultiHeadAttention(
        num_heads=4,
        key_dim=16
    )(x, x)

    x = Add()([x, attention])

    x = LayerNormalization()(x)

    x = Bidirectional(
        LSTM(32)
    )(x)

    x = Dense(128, activation="relu")(x)

    x = Dropout(0.3)(x)

    outputs = Dense(
        num_classes,
        activation="softmax"
    )(x)

    return Model(inputs, outputs)


# ============================================================
# 10. FULL TO–TBi-STMNet MODEL
# ============================================================

def build_full_model(input_shape):

    inputs = Input(shape=input_shape)

    # Bidirectional temporal representation
    x = Bidirectional(
        LSTM(64, return_sequences=True)
    )(inputs)

    # Transformer attention
    attention = MultiHeadAttention(
        num_heads=4,
        key_dim=16
    )(x, x)

    x = Add()([x, attention])

    x = LayerNormalization()(x)

    # Deep bidirectional learning
    x = Bidirectional(
        LSTM(32)
    )(x)

    x = Dense(128, activation="relu")(x)

    x = Dropout(0.25)(x)

    x = Dense(64, activation="relu")(x)

    outputs = Dense(
        num_classes,
        activation="softmax"
    )(x)

    return Model(inputs, outputs)


# ============================================================
# 11. RUN ABLATION EXPERIMENT
# ============================================================

def run_ablation(name, X, model_type="full"):

    print("\n" + "-" * 90)
    print("Running:", name)
    print("-" * 90)

    X_train_seq, X_test_seq = prepare_data(X)

    y_train = y_ablation[train_idx]
    y_test = y_ablation[test_idx]

    input_shape = X_train_seq.shape[1:]

    # Model selection
    if model_type == "transformer":
        model = build_transformer_only(input_shape)

    elif model_type == "bilstm":
        model = build_bilstm_only(input_shape)

    elif model_type == "transformer_bilstm":
        model = build_transformer_bilstm(input_shape)

    else:
        model = build_full_model(input_shape)

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    start_time = time.time()

    model.fit(
        X_train_seq,
        y_train,
        epochs=3,
        batch_size=32,
        validation_split=0.10,
        callbacks=[
            EarlyStopping(
                monitor="val_loss",
                patience=5,
                restore_best_weights=True
            )
        ],
        verbose=0
    )

    training_time = time.time() - start_time

    predictions = np.argmax(
        model.predict(X_test_seq, verbose=0),
        axis=1
    )

    accuracy = accuracy_score(
        y_test,
        predictions
    ) * 100

    ablation_results.append({
        "Ablation configurations": name,
        "Accuracy (%)": accuracy,
        "Training Time (s)": training_time
    })


# ============================================================
# 12. IMAGE-ONLY
# ============================================================

run_ablation(
    "Image-only",
    X_image,
    model_type="full"
)


# ============================================================
# 13. TEXT-ONLY
# ============================================================

run_ablation(
    "Text-only",
    X_text,
    model_type="full"
)


# ============================================================
# 14. MULTIMODAL FUSION
# ============================================================

run_ablation(
    "Multimodal fusion",
    X_multimodal,
    model_type="full"
)


# ============================================================
# 15. TRANSFORMER-ONLY
# ============================================================

run_ablation(
    "Transformer-only",
    X_multimodal,
    model_type="transformer"
)


# ============================================================
# 16. BiLSTM-ONLY
# ============================================================

run_ablation(
    "BiLSTM-only",
    X_multimodal,
    model_type="bilstm"
)


# ============================================================
# 17. TRANSFORMER + BiLSTM
# ============================================================

run_ablation(
    "Transformer + BiLSTM",
    X_multimodal,
    model_type="transformer_bilstm"
)


# ============================================================
# 18. FULL PROPOSED MODEL
# ============================================================

run_ablation(
    "Full model (TO–TBi-STMNet)",
    X_multimodal,
    model_type="full"
)


# ============================================================
# 19. DISPLAY ABLATION TABLE
# ============================================================

ablation_df = pd.DataFrame(ablation_results)

print("\n")
print("=" * 90)
print("ABLATION STUDY RESULTS")
print("=" * 90)

print(
    ablation_df[
        [
            "Ablation configurations",
            "Accuracy (%)"
        ]
    ].to_string(
        index=False,
        formatters={
            "Accuracy (%)": "{:.2f}".format
        }
    )
)


# ============================================================
# 20. SAVE RESULTS
# ============================================================

ablation_df.to_csv(
    "TO_TBi_STMNet_ablation_results.csv",
    index=False
)

print("\nSaved: TO_TBi_STMNet_ablation_results.csv")

print("\nABLATION STUDY COMPLETED SUCCESSFULLY")
print("=" * 90)
```

```python

# ============================================================
# STEP 11: 5-RUN STATISTICAL SIGNIFICANCE ANALYSIS
# TO–TBi-STMNet vs P2CatBoost
# ============================================================

import numpy as np
import pandas as pd
import tensorflow as tf
import gc
import warnings

from scipy.stats import ttest_rel, t

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score
)

from catboost import CatBoostClassifier

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    LSTM,
    Bidirectional,
    MultiHeadAttention,
    LayerNormalization,
    Add
)
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")


# ============================================================
# 1. CONFIGURATION
# ============================================================

N_RUNS = 5
TEST_SIZE = 0.20
BASE_SEED = 42

EPOCHS = 3
BATCH_SIZE = 32

METRICS = [
    "Accuracy",
    "Precision",
    "Recall"
]


# ============================================================
# 2. PREPARE DATA
# ============================================================

X = np.asarray(fused_features, dtype=np.float32)
y_raw = np.asarray(y).ravel()

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y_raw)

NUM_CLASSES = len(np.unique(y_encoded))


# ============================================================
# 3. PROPOSED MODEL
# ============================================================

def build_proposed_model(input_shape, num_classes):

    inputs = Input(shape=input_shape)

    # BiLSTM representation
    x = Bidirectional(
        LSTM(64, return_sequences=True)
    )(inputs)

    # Transformer attention
    attention = MultiHeadAttention(
        num_heads=4,
        key_dim=16
    )(x, x)

    x = Add()([x, attention])

    x = LayerNormalization()(x)

    # Bidirectional sequence learning
    x = Bidirectional(
        LSTM(32)
    )(x)

    x = Dense(128, activation="relu")(x)

    x = Dropout(0.25)(x)

    x = Dense(64, activation="relu")(x)

    outputs = Dense(
        num_classes,
        activation="softmax"
    )(x)

    model = Model(
        inputs,
        outputs,
        name="TO_TBi_STMNet"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.001
        ),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ============================================================
# 4. METRIC CALCULATION
# ============================================================

def calculate_metrics(y_true, y_pred):

    return {
        "Accuracy": accuracy_score(
            y_true,
            y_pred
        ),

        "Precision": precision_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        ),

        "Recall": recall_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )
    }


# ============================================================
# 5. RUN STORAGE
# ============================================================

proposed_runs = {
    "Accuracy": [],
    "Precision": [],
    "Recall": []
}

catboost_runs = {
    "Accuracy": [],
    "Precision": [],
    "Recall": []
}


# ============================================================
# 6. FIVE INDEPENDENT RUNS
# ============================================================

for run in range(N_RUNS):

    seed = BASE_SEED + run

    print("\n" + "=" * 100)
    print(f"RUN {run + 1}/{N_RUNS} | RANDOM SEED = {seed}")
    print("=" * 100)

    np.random.seed(seed)
    tf.random.set_seed(seed)

    # --------------------------------------------------------
    # Train-test split
    # --------------------------------------------------------

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y_encoded,
        test_size=TEST_SIZE,
        stratify=y_encoded,
        random_state=seed
    )

    # --------------------------------------------------------
    # Scaling
    # --------------------------------------------------------

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(
        X_train
    ).astype(np.float32)

    X_test_scaled = scaler.transform(
        X_test
    ).astype(np.float32)

    # Sequence representation for deep model
    X_train_seq = X_train_scaled.reshape(
        X_train_scaled.shape[0],
        X_train_scaled.shape[1],
        1
    )

    X_test_seq = X_test_scaled.reshape(
        X_test_scaled.shape[0],
        X_test_scaled.shape[1],
        1
    )

    # ========================================================
    # 7. P2CatBoost
    # ========================================================

    print("\nTraining P2CatBoost...")

    cat_model = CatBoostClassifier(
        iterations=5,
        depth=8,
        learning_rate=0.05,
        loss_function="MultiClass",
        random_seed=seed,
        verbose=False,
        allow_writing_files=False
    )

    cat_model.fit(
        X_train_scaled,
        y_train
    )

    cat_pred = cat_model.predict(
        X_test_scaled
    ).ravel().astype(int)

    cat_metrics = calculate_metrics(
        y_test,
        cat_pred
    )

    for metric in METRICS:
        catboost_runs[metric].append(
            cat_metrics[metric]
        )


    # ========================================================
    # 8. TO–TBi-STMNet
    # ========================================================

    print("\nTraining TO–TBi-STMNet...")

    tf.keras.backend.clear_session()
    tf.random.set_seed(seed)

    proposed_model = build_proposed_model(
        input_shape=X_train_seq.shape[1:],
        num_classes=NUM_CLASSES
    )

    proposed_model.fit(
        X_train_seq,
        y_train,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        validation_split=0.10,
        shuffle=True,
        callbacks=[
            EarlyStopping(
                monitor="val_loss",
                patience=5,
                restore_best_weights=True
            )
        ],
        verbose=0
    )

    proposed_prob = proposed_model.predict(
        X_test_seq,
        verbose=0
    )

    proposed_pred = np.argmax(
        proposed_prob,
        axis=1
    )

    proposed_metrics = calculate_metrics(
        y_test,
        proposed_pred
    )

    for metric in METRICS:
        proposed_runs[metric].append(
            proposed_metrics[metric]
        )


    # --------------------------------------------------------
    # Release memory
    # --------------------------------------------------------

    del proposed_model
    del cat_model

    tf.keras.backend.clear_session()
    gc.collect()


# ============================================================
# 9. STATISTICAL ANALYSIS
# ============================================================

statistical_results = []


for metric in METRICS:

    proposed_values = np.array(
        proposed_runs[metric],
        dtype=np.float64
    )

    cat_values = np.array(
        catboost_runs[metric],
        dtype=np.float64
    )

    # --------------------------------------------------------
    # Mean and standard deviation
    # --------------------------------------------------------

    proposed_mean = proposed_values.mean()

    proposed_sd = proposed_values.std(ddof=1)

    cat_mean = cat_values.mean()

    cat_sd = cat_values.std(ddof=1)

    # --------------------------------------------------------
    # Paired difference
    # Proposed - CatBoost
    # --------------------------------------------------------

    differences = proposed_values - cat_values

    mean_difference = differences.mean()

    sd_difference = differences.std(ddof=1)

    # --------------------------------------------------------
    # Paired t-test
    # --------------------------------------------------------

    t_value, p_value = ttest_rel(
        proposed_values,
        cat_values
    )

    df = N_RUNS - 1

    # --------------------------------------------------------
    # 95% Confidence Interval
    # --------------------------------------------------------

    t_critical = t.ppf(
        0.975,
        df
    )

    standard_error = (
        sd_difference / np.sqrt(N_RUNS)
    )

    ci_lower = (
        mean_difference
        - t_critical * standard_error
    )

    ci_upper = (
        mean_difference
        + t_critical * standard_error
    )

    # --------------------------------------------------------
    # Cohen's d for paired samples (dz)
    # --------------------------------------------------------

    if sd_difference != 0:

        cohens_d = (
            mean_difference / sd_difference
        )

    else:

        cohens_d = np.nan

    # --------------------------------------------------------
    # Statistical significance
    # --------------------------------------------------------

    significant = (
        "significant"
        if p_value < 0.05
        else "Not significant"
    )

    # --------------------------------------------------------
    # Store
    # --------------------------------------------------------

    statistical_results.append({

        "Performance Metrics": metric,

        "TO–TBi-STMNet (Mean ± SD)": (
            f"{proposed_mean:.4f} ± {proposed_sd:.4f}"
        ),

        "P2CatBoost (Mean ± SD)": (
            f"{cat_mean:.4f} ± {cat_sd:.4f}"
        ),

        "Mean Difference (95% CI)": (
            f"{mean_difference:.4f} "
            f"[{ci_lower:.4f}, {ci_upper:.4f}]"
        ),

        "t-value (df = 4)": t_value,

        "p-value": p_value,

        "Cohen's d": cohens_d,

        "Significant": significant

    })


# ============================================================
# 10. FINAL STATISTICAL TABLE
# ============================================================

stat_df = pd.DataFrame(
    statistical_results
)

print("\n")
print("=" * 170)
print("5-RUN PAIRED STATISTICAL COMPARISON")
print("=" * 170)

print(
    stat_df.to_string(
        index=False,
        formatters={
            "t-value (df = 4)": "{:.4f}".format,
            "p-value": "{:.5f}".format,
            "Cohen’s d": "{:.4f}".format
        }
    )
)


# ============================================================
# 12. SAVE STATISTICAL RESULTS
# ============================================================

stat_df.to_csv(
    "TO_TBi_STMNet_statistical_comparison.csv",
    index=False
)

accuracy_runs_df.to_csv(
    "TO_TBi_STMNet_five_run_accuracy.csv",
    index=False
)

print("\nSaved files:")
print("1. TO_TBi_STMNet_statistical_comparison.csv")
print("2. TO_TBi_STMNet_five_run_accuracy.csv")
```

```python

# ============================================================
# STEP 12: 5-RUN PERFORMANCE EVALUATION
# TO–TBi-STMNet (PROPOSED)
# ============================================================

import numpy as np
import pandas as pd
import tensorflow as tf
import warnings

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    LSTM,
    Bidirectional,
    MultiHeadAttention,
    LayerNormalization,
    Add
)
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")


# ============================================================
# 1. CONFIGURATION
# ============================================================

RANDOM_SEEDS = [12, 24, 36, 42, 50]

EPOCHS = 2
BATCH_SIZE = 32

print("=" * 100)
print("5-RUN PERFORMANCE EVALUATION")
print("TO–TBi-STMNet [PROPOSED]")
print("=" * 100)


# ============================================================
# 2. PREPARE FEATURES AND TARGET
# ============================================================

X = np.asarray(fused_features, dtype=np.float32)

y_raw = np.asarray(y).ravel()

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y_raw)

NUM_CLASSES = len(np.unique(y_encoded))

print("\nFused Feature Shape :", X.shape)
print("Target Shape       :", y_encoded.shape)
print("Number of Classes  :", NUM_CLASSES)


# ============================================================
# 3. MODEL ARCHITECTURE
# ============================================================

def build_to_tbi_stmnet(input_shape, num_classes):

    inputs = Input(shape=input_shape)

    # Bidirectional LSTM feature representation
    x = Bidirectional(
        LSTM(64, return_sequences=True)
    )(inputs)

    # Transformer attention block
    attention = MultiHeadAttention(
        num_heads=4,
        key_dim=16
    )(x, x)

    x = Add()([x, attention])

    x = LayerNormalization()(x)

    # Bidirectional sequence learning
    x = Bidirectional(
        LSTM(32)
    )(x)

    x = Dense(
        128,
        activation="relu"
    )(x)

    x = Dropout(0.25)(x)

    x = Dense(
        64,
        activation="relu"
    )(x)

    outputs = Dense(
        num_classes,
        activation="softmax"
    )(x)

    model = Model(
        inputs,
        outputs,
        name="TO_TBi_STMNet"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.001
        ),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ============================================================
# 4. STORAGE FOR RUN RESULTS
# ============================================================

run_results = []


# ============================================================
# 5. EXECUTE FIVE RUNS
# ============================================================

for run_number, seed in enumerate(RANDOM_SEEDS, start=1):

    print("\n" + "=" * 100)
    print(f"RUN {run_number} | RANDOM SEED = {seed}")
    print("=" * 100)

    # Set random seeds
    np.random.seed(seed)
    tf.random.set_seed(seed)

    # Clear previous model session
    tf.keras.backend.clear_session()

    # --------------------------------------------------------
    # Train-test split
    # --------------------------------------------------------

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y_encoded,
        test_size=0.20,
        random_state=seed,
        stratify=y_encoded
    )

    # --------------------------------------------------------
    # Feature scaling
    # --------------------------------------------------------

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(
        X_train
    ).astype(np.float32)

    X_test_scaled = scaler.transform(
        X_test
    ).astype(np.float32)

    # --------------------------------------------------------
    # Reshape for sequential model
    # --------------------------------------------------------

    X_train_seq = X_train_scaled.reshape(
        X_train_scaled.shape[0],
        X_train_scaled.shape[1],
        1
    )

    X_test_seq = X_test_scaled.reshape(
        X_test_scaled.shape[0],
        X_test_scaled.shape[1],
        1
    )

    input_shape = X_train_seq.shape[1:]

    # --------------------------------------------------------
    # Build model
    # --------------------------------------------------------

    model = build_to_tbi_stmnet(
        input_shape,
        NUM_CLASSES
    )

    # --------------------------------------------------------
    # Train model
    # --------------------------------------------------------

    model.fit(
        X_train_seq,
        y_train,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        validation_split=0.10,
        callbacks=[
            EarlyStopping(
                monitor="val_loss",
                patience=7,
                restore_best_weights=True
            )
        ],
        verbose=0
    )

    # --------------------------------------------------------
    # Prediction
    # --------------------------------------------------------

    probabilities = model.predict(
        X_test_seq,
        verbose=0
    )

    predictions = np.argmax(
        probabilities,
        axis=1
    )

    # --------------------------------------------------------
    # Calculate metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_test,
        predictions
    ) * 100

    precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    ) * 100

    recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    ) * 100

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    ) * 100

    # --------------------------------------------------------
    # Store run result
    # --------------------------------------------------------

    run_results.append({
        "Runs": f"Run {run_number}",
        "Random Seed": seed,
        "Accuracy (%)": accuracy,
        "Precision (%)": precision,
        "Recall (%)": recall,
        "F1-score (%)": f1
    })

    # --------------------------------------------------------
    # Display current run
    # --------------------------------------------------------

    print(f"Accuracy  : {accuracy:.3f}%")
    print(f"Precision : {precision:.3f}%")
    print(f"Recall    : {recall:.3f}%")
    print(f"F1-score : {f1:.3f}%")

    del model


# ============================================================
# 6. CREATE FINAL TABLE
# ============================================================

results_df = pd.DataFrame(run_results)


# ============================================================
# 7. PRINT FINAL 5-RUN TABLE
# ============================================================

print("\n")
print("=" * 115)
print("TO–TBi-STMNet - 5 RUN PERFORMANCE RESULTS")
print("=" * 115)

print(
    results_df.to_string(
        index=False,
        formatters={
            "Accuracy (%)": "{:.2f}".format,
            "Precision (%)": "{:.2f}".format,
            "Recall (%)": "{:.2f}".format,
            "F1-score (%)": "{:.2f}".format
        }
    )
)


# ============================================================
# 9. SAVE RESULTS
# ============================================================

results_df.to_csv(
    "TO_TBi_STMNet_5Run_Performance.csv",
    index=False
)

print("\nSaved: TO_TBi_STMNet_5Run_Performance.csv")


# ============================================================
# 10. COMPLETION
# ============================================================

print("\n")
print("=" * 100)
print("5-RUN EVALUATION COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ============================================================
# SHAP BEESWARM PLOT FOR text_data.csv
# FAST VERSION
# ============================================================

import pandas as pd
import numpy as np
import shap
import matplotlib.pyplot as plt

from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
from sklearn.ensemble import ExtraTreesClassifier

# ============================================================
# 1. LOAD DATASET
# ============================================================

file_path = r"E:\Projects\2_Enterprises\text_data.csv"

df = pd.read_csv(file_path)

# ============================================================
# 2. TARGET COLUMN
# ============================================================

# Assumes the target is the last column
TARGET_COLUMN = df.columns[-1]

X = df.drop(columns=[TARGET_COLUMN]).copy()
y = df[TARGET_COLUMN].copy()


# ============================================================
# 3. HANDLE FEATURE VALUES
# ============================================================

# Convert text columns into numeric representations
for col in X.columns:

    if X[col].dtype == "object" or str(X[col].dtype) == "category":

        X[col] = X[col].fillna("Unknown").astype(str)

        encoder = OrdinalEncoder(
            handle_unknown="use_encoded_value",
            unknown_value=-1
        )

        X[col] = encoder.fit_transform(X[[col]]).ravel()

    else:

        X[col] = X[col].fillna(X[col].median())

X = X.replace([np.inf, -np.inf], 0)

X = X.astype(np.float32)

# Encode target labels
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y.astype(str))

# ============================================================
# 4. REDUCE DATA SIZE FOR SPEED
# ============================================================

sample_size = min(2000, len(X))

rng = np.random.RandomState(42)

indices = rng.choice(
    len(X),
    sample_size,
    replace=False
)

X_sample = X.iloc[indices].values
y_sample = y[indices]

feature_names = X.columns.tolist()

# ============================================================
# 5. TRAIN FAST TREE MODEL
# ============================================================

print("\nTraining SHAP model...")

model = ExtraTreesClassifier(
    n_estimators=40,
    max_depth=10,
    min_samples_leaf=3,
    max_features="sqrt",
    n_jobs=-1,
    random_state=42
)

model.fit(X_sample, y_sample)

print("Model training completed.")

# ============================================================
# 6. COMPUTE SHAP VALUES
# ============================================================

print("Calculating SHAP values...")

explainer = shap.TreeExplainer(model)

shap_values = explainer.shap_values(X_sample)

# ============================================================
# 7. MULTICLASS SHAP HANDLING
# ============================================================

if isinstance(shap_values, list):

    # Average absolute importance across classes
    shap_plot_values = np.mean(
        np.stack([np.abs(sv) for sv in shap_values], axis=0),
        axis=0
    )

else:

    shap_values = np.asarray(shap_values)

    if shap_values.ndim == 3:

        # Samples x features x classes
        shap_plot_values = np.mean(
            np.abs(shap_values),
            axis=2
        )

    else:

        shap_plot_values = shap_values

# ============================================================
# 8. SHAP BEESWARM PLOT
# ============================================================

plt.figure(figsize=(12, 8))

shap.summary_plot(
    shap_plot_values,
    X_sample,
    feature_names=feature_names,
    max_display=20,
    plot_type="dot",
    show=False
)

plt.title(
    "SHAP Beeswarm Plot – Text Dataset",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

# ============================================================
# 9. SAVE FIGURE
# ============================================================

plt.show()

print("\nSHAP Beeswarm generated successfully!")
print("Saved as: TextData_SHAP_Beeswarm.png")
```
